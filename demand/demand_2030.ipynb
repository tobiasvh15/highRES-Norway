```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
demand_base = pd.read_csv('europe_demand_2006-2015.csv', parse_dates=['datetime']).set_index('datetime')[['NO03', 'NO11', 'NO15', 'NO18', 'NO30', 'NO34', 'NO38', 'NO42', 'NO46', 'NO50', 'NO54']]/1000 #GWh
```

```python
demand_base #[demand_base.index.year==2005]
```

```python
average_demand = demand_base.sum().sum()/(demand_base.index.year.to_series().unique().size)
average_demand/1000 #TWh
```

```python
demand_scaled = demand_base*(140000/average_demand)
```

```python
demand = demand_scaled.copy()
```

```python
print(demand.sum().sum()/(demand_base.index.year.to_series().unique().size*1000)) # calculate average  #TWh
demand
```

# Plot

Base and 2030 yearly average demand

```python
yearly_demand = []

for i in range(1951, 2022):
    yearly_demand.append(demand_base[demand_base.index.year == i].sum().sum()/1000)
```

```python
average_demand = sum(yearly_demand)/len(yearly_demand)
average_demand
```

```python
plt.plot(range(1951, 2022), yearly_demand, '-o', label='Yearly demand')
plt.plot([1951, 2021], [average_demand, average_demand], '--', color='red', label='Average')
plt.legend()
plt.show()
```

```python
yearly_demand_2030 = []

for i in range(1951, 2022):
    yearly_demand_2030.append(demand[demand.index.year == i].sum().sum()/1000)
```

```python
plt.plot(range(1951, 2022), yearly_demand, '-o', color='orange')
plt.plot(range(1951, 2022), yearly_demand_2030, '-o', color='green')
plt.plot([1951, 2021], [sum(yearly_demand)/len(yearly_demand), sum(yearly_demand)/len(yearly_demand)], '--', color='red', label='Average')
plt.plot([1951, 2021], [sum(yearly_demand_2030)/len(yearly_demand_2030), sum(yearly_demand_2030)/len(yearly_demand_2030)], '--', color='red', label='Average')
plt.legend(['Base', '2030', 'Average'])
plt.xlabel('Year')
plt.ylabel('Demand (TWh)')
plt.show()
```

```python
max(yearly_demand_2030)
```
