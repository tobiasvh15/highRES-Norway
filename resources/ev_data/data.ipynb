```python
import pandas as pd
import numpy as np
```

```python
demand_driving = pd.read_csv('demand_driving_MWh.tsv', sep='\t', header=None)
```

```python
demand_driving
```

```python
demand_driving['zone'] = demand_driving[0].str[-4:]
demand_driving['hour'] = demand_driving[0].str[:-5]
demand_driving = demand_driving.drop(columns=[0]).rename(columns={1:'value'})
```

```python
demand_driving
```

```python
avg = {}
for i in demand_driving.zone.unique():
    avg[i] = (demand_driving[demand_driving.zone == i].value.sum()/365).round(6)
```

```python
np.mean(list(avg.values()))*1000 #kWh
```
