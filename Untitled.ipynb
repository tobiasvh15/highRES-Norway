```python
%%time
import concurrent.futures
import pathlib

import netCDF4 as nc
import numpy as np

# List of netCDF files
files = pathlib.Path("/cluster/work/projects/ec85/SARAH/2010/SID/").rglob("SI*.nc")

# Variable name
var_name = "record_status"  # Change it to your variable name

# Set to store unique values
unique_values = set()


def process_file(file):
    # Open netCDF file for reading
    with nc.Dataset(file, "r") as dataset:
        # Get variable data
        var_data = dataset.variables[var_name][:]

        # Extract unique values from the variable data
        return np.unique(var_data)


# Create a ThreadPoolExecutor
with concurrent.futures.ThreadPoolExecutor() as executor:
    # Submit the processing tasks for each file
    future_results = [executor.submit(process_file, file) for file in files]

    # Loop over the completed tasks to get the results
    for future in concurrent.futures.as_completed(future_results):
        unique_values.update(future.result())

# Print the unique values
print("Unique values:", unique_values)
```

```python
import concurrent.futures
import pathlib

import netCDF4 as nc
import numpy as np

# List of netCDF files
files = pathlib.Path("/cluster/work/projects/ec85/SARAH/2010/SIS/").rglob("SI*.nc")

# Variable name
var_name = "record_status"  # Change it to your variable name

# Dictionary to store files associated with each unique value
value_files = {}


def process_file(file):
    # Open netCDF file for reading
    with nc.Dataset(file, "r") as dataset:
        # Get variable data
        var_data = dataset.variables[var_name][:]

        # Extract unique values from the variable data
        unique_values = np.unique(var_data)

        # Update the value_files dictionary
        for value in unique_values:
            if value in value_files:
                value_files[value].append(file)
            else:
                value_files[value] = [file]


# Create a ThreadPoolExecutor
with concurrent.futures.ThreadPoolExecutor() as executor:
    # Submit the processing tasks for each file
    future_results = [executor.submit(process_file, file) for file in files]

    # Wait for all tasks to complete
    concurrent.futures.wait(future_results)

# Print the files associated with each unique value
for value, files in value_files.items():
    print(f"Value {value} is present in files: {files}")
```
